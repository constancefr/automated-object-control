```python
import tensorflow as tf
import tf2onnx
import os
```

```python
already_generate_idx = False
already_generate_idx = True

if not already_generate_idx:
    model = tf.keras.models.load_model('nn_model.keras')
    model.summary()
    
    input_signature = [
        tf.TensorSpec(inp.shape, inp.dtype, name=inp.name) for inp in model.inputs
    ]

    onnx_model, _ = tf2onnx.convert.from_keras(
        model, 
        input_signature=input_signature, 
        opset=13
    )
    with open('nn_model.onnx', "wb") as f:
        f.write(onnx_model.SerializeToString())
```
